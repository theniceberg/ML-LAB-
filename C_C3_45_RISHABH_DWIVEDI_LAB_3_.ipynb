```python
# Separate features (X) and target (y)
X = data.iloc[:, :-1]
y = data.iloc[:, -1]
```

```python
def find_s(X,y):
    hypo= None
    print("Intial Hypothesis:",hypo)
    for i in range(len(X)):
        if y[i] == "Yes":
            if hypo is None:
                hypo=X.iloc[i].values.copy()
            else:
                for j in range(len(hypo)):
                    if hypo[j] != X.iloc[i].values[j]:
                        hypo[j]="?"
        print(f"\nAfter Training Example {i+1}")
        print(hypo)
    return hypo
final_hypo=find_s(X,y)
print("\nFinal Hypothesis",final_hypo)
```

```python
import copy
def candidate_elimination(concepts, target):
  specific = concepts.iloc[0].values.copy()
  general = [["?" for _ in range(len(specific))]]

  print("Intial Specific Hypothesis:",specific)
  print("Intial General Hypothesis:",general)


  for i in range(len(concepts)):
    h = concepts.iloc[i].values
    current_target = target.iloc[i]

    if current_target == "Yes":
      for x in range(len(specific)):
        if h[x] != specific[x]:
          specific[x] = "?"

    else:
      for x in range(len(specific)):
        if h[x] != specific[x]:
          general.append(["?" if j != x else specific[x] for j in range(len(specific))])


    print(f"\nExample {i+1}")
    print("Specific =",specific)
    print("General =",general)

  return specific,general
S,G = candidate_elimination(X,y)
print("\nInitial Specific boundary")
print(S)
print("Final General Hypothesis:")
print(G)
```
